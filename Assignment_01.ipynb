```python
#create dictionary with 5 elements, and perform CRUD operations on it
my_dict={101:"Rahul",102:"Aman",103:"Sanya",104:"Priya",105:"Priya"}
print("students information:", my_dict)
#Add student detail
my_dict[106]="Neha"
print("After adding a students detail=",my_dict)
#Delete student information
del my_dict[102]
print("After deleting a students information=",my_dict)
#Display student information
print("final student information=")
for roll_no,name in my_dict.items():
    print(roll_no,name)
```

```python
#create two lists and zip them together to form the dictionary
list1=['name','panel','roll_no']
list2=['ABC','B',34]
my_dict=dict(zip(list1,list2))
print("Dictionary =",my_dict)
```

```python
#using sorted() function
my_dict={'name':'ABC','panel':'B','roll_no':34,'marks':[65,87,67,94]}
print(sorted(my_dict))
```

```python
#creating two list from a Dictionary, one of keys, and one of values
my_dict={'name':'ABC','panel':'B','roll_no':34,'marks':[65,87,67,94]}
keys=list(my_dict.keys())
values=list(my_dict.values())
print('keys=',keys)
print('values=',values)
```

```python
#find mean values of a dictionary
my_dict={'marks1':23,'marks2':123,'marks3':34,'marks4':13,'marks5':39}
values=my_dict.values()
mean=sum(values)/len(values)
print("Mean of all values=",mean)
```

```python
#perform given operation on dictionary
my_dict={'name':['Yash','Neel','Dev'],'roll_no':[11,12,13],'marks':[98,56,98]}
#dispaly name a Dev
print(my_dict['name'][2])
#Display roll no 12
print(my_dict['roll_no'][1])
#Display greatest marks
print(max(my_dict['marks']))
```

```python
#calculate frequency of lettters occuring in a given string
string=("hello world!")
freq={}
for i in string:
    if i in freq:
        freq[i]+=1
    else:
        freq[i]=1
print(freq)
```

```python
 #Calculate the number of days between two given dates
from datetime import date
d1=int(input("enter day of first date="))
m1=int(input("enter month of first date="))
y1=int(input("enter year of first date="))
d2=int(input("enter day of second date="))
m2=int(input("enter month of second date="))
y2=int(input("enter year of second date="))
date1=date(y1,m1,d1)
date2=date(y2,m2,d2)
days=abs((date2-date1).days)
print("Number of days=",days)
```
