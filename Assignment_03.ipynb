```python
#to check whether the triangle is right angled or not
def right_triangle(a,b,c):
    #check which side is the largest
    if a>=b and a>=c:
        largest=a
        side1=b
        side2=c
    elif b>=a and b>=c:
        largest=b
        side1=a
        side2=c
    else:
        largest=c
        side1=a
        side2=b
    #check pythagorus theorem
    if side1**2+side2**2==largest**2:
        return True
    else:
        return False
#input three sides
a=float(input("enter first side="))
b=float(input("enter second side="))
c=float(input("enter third side="))
if right_triangle(a,b,c):
    print("the triangle is a right-angled triangle.")
else:
    print("the triangle is NOT a right-angled triangle.")
```

```python
#display prime factor of given number
num=int(input("enter a number="))
print("prime factor are=",end=" ")
factor=2
while num>1:
    if num%factor==0:
        print(factor,end=" ")
        num=num//factor
    else:
        factor+=1
```

```python
#print all even numbers from 1 to 100
def print_even_numbers():
    for i in range(1,101):
        if i%2==0:
            print(i,end=" ")
print_even_numbers()
```
