```python
#Take three numbers from user
a=float(input("enter first number="))
b=float(input("enter second number="))
c=float(input("enter third number="))
#finding the largest of three
if a>=b and a>=c:
    largest=a
elif b>=a and b>=c:
    largest=b
else:
    largest=c
print("largest of three numbers is=",largest)
```
